# Exploration

Point `CONFIG` at the experiment YAML; the data paths, target and columns come from it. Run from the repository root (or let the first cell move there).

```python
import os, pathlib
if pathlib.Path.cwd().name == 'notebooks':
    os.chdir('..')

from mlt.config import load_config, targets
from mlt.data import feature_columns, load_data, resolve_task
from mlt import eda

CONFIG = 'configs/examples/breast_cancer.yaml'
cfg = load_config(CONFIG)
train, test, sample = load_data(cfg)
task, metric = resolve_task(cfg, train)
target = targets(cfg)[0]
features, cats, nums, dates = feature_columns(cfg, train, test)
print(task, metric, train.shape, None if test is None else test.shape)
```

## Columns: types, missing values, presence in test

```python
eda.overview(train, test)
```

## Target

```python
train[target].value_counts(normalize=True) if task != 'regression' else train[target].describe()
```

## Univariate statistics

```python
eda.univariate_stats(train[nums + cats])
```

## Train vs test distributions (a gap here means the CV may not track the leaderboard)

```python
eda.plot_distributions(train, test, nums[:12]);
```

```python
eda.plot_counts(train, cats[:6]) if cats else None;
```

## Correlations and association with the target

```python
eda.plot_corr(train, nums[:25]);
```

```python
eda.bivariate_stats(train[nums + cats + [target]], target).head(20)
```

## Class structure in 2-D (t-SNE on numeric columns)

```python
eda.plot_tsne(train[nums].to_numpy(), train[target], n=1500);
```
